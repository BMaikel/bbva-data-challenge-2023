# BBVA Data Challenge 2023

## Introducción:

El objetivo del reto es desarrollar un modelo de Machine Learning capaz de predecir el motivo por el cual un cliente abandonará el segmento objetivo del banco.

El problema corresponde a una tarea de clasificación multiclase, donde cada cliente pertenece a una de cinco categorías de abandono.

La solución será evaluada utilizando el Macro F1-Score, priorizando un buen desempeño sobre todas las clases.

## Comprensión del problema:

| **Aspecto**        | **Descripción**                                 |
|--------------------|-------------------------------------------------|
| Organización       | BBVA                                            |
| Dominio            | Banca                                           |
| Problema           | Identificar clientes que abandonan el segmento. |
| Impacto            | Reducir pérdida de clientes valiosos.           |
| Beneficio Esperado | Permitir campañas de retención.                 |

Si el banco identifica anticipadamente el motivo del abandono, puede implementar estrategias específicas para reducir la pérdida de clientes y mejorar su rentabilidad.

## Objetivos del negocio

### Objetivo General
Identificar oportunamente el motivo de abandono de los clientes para facilitar acciones preventivas por parte del banco.

### Objetivos Específicos
- Detectar clientes con alto riesgo
- Identificar el motivo del abandono
- Identificar variables relacionadas con la deserción de clientes.

## Métrica

Debido a que el problema es multiclase y puede existir desbalance entre las categorías, la competencia utiliza el Macro F1-score, el cual calcula el F1 de cada clase y posteriormente obtiene el promedio simple entre ellas.
